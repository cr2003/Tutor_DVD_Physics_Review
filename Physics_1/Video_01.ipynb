###  Velocity and Acceleration in One Dimension

#### Displacement
The movement from initial to final position

$$\Delta x = x_f - x_i$$

Units in meters

$$\Delta x = [m]$$

#### Velocity
Units in meters per second

$$velocity = [m/s] $$

#### Average velocity
Units in meters per second

$$\bar{v} = \frac{\Delta x}{\Delta t} = \frac{x_f - x_i}{t_f - t_i} = \left[\frac{m}{s}\right] $$

🦉 **Note (crv):**

We are going to use var_qty to denote a variable for AstroPy with astro units.

#### Problem
A person runs a marathon in 2h, 9 min, 21 s. The marathon distance is 26 mi, 385 yd.

Calculate the average speed in mile per hour.

```python
# dt, dx (delta time, delta x)
dt_qty = 2 * u.h + 9 * u.min + 21 * u.s
dx_qty = 26 * u.imperial.mi + 385 * u.imperial.yd

avg_v_qty = dx_qty / dt_qty
print(f"Average speed: {avg_v_qty:.1f}")
```

#### Acceleration

$$\bar{a} = \frac{\Delta v}{\Delta t} = \frac{v_f - v_i}{t_f - t_i} = \left[\frac{m/s}{s} \right] =  \left[\frac{m}{s^2}\right]$$


#### Problem
We are sending a man to the moon by shooting him out of a canon. The canon is 220 m long and velocity of the person as he exits the canon is 10.97 km/s. What acceleration does he experience as he travels down the canon?

<img src="../pics/Phy1_video_1.png" width=400> 


```python
# --- SymPy symbol vars ---
a, v, v0, x = sp.symbols("a, v, v_0, x", real=True, positive=True)

# --- kinematic equation ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * x)
a_expr = sp.solve(eq_vel_sq, a)

# --- results ---
print("From the equation:")
display(Math(rf"{sp.latex(eq_vel_sq)}"))
print("Solve for a:")
display(Math(rf"a = {sp.latex(a_expr)}"))
```

```python
# --- 1. Create Lambdified Functions ---
calc_a = sp.lambdify((v, v0, x), a_expr)

# --- 2. Astropy Physical Inputs ---
x_qty = 220 * u.m
v_qty = (10.97 * u.km / u.s).to(u.m / u.s)
v0_qty = 0 * (u.m / u.s)

# --- 3. Numerical Calculations ---
a_qty = calc_a(v_qty, v0_qty, x_qty)[0]
g_factor_qty = a_qty / const.g0

# --- 4. Results ---
print(f"Acceleration: {a_qty:,.0f} equivalent to {a_qty.to(u.km / u.s**2):,.0f}")
print(f"The g factor is: {g_factor_qty:,.0f} g")
```
