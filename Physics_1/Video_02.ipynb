##  Equations of Motion in 1 Dimension

### Motion with constant acceleration

$$ x = v_0 t + \frac{1}{2}at^2 $$

$$v = v_0 + at$$

$$v^2 = v_0^2 + 2ax$$

#### Problem 1

A car is traveling at 60 mi/hr and the driver sees an deer 100m away on the highway.

What acceleration is needed for the car to stop before hitting the deer?

<img src="../pics/Phy1_video_2.png" width=600> 

```python
# --- SymPy symbol vars ---
a, v, v0, x = sp.symbols("a, v, v_0, x", real=True, positive=True)

# --- kinematic equation ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * x)
a_expr = sp.solve(eq_vel_sq, a)[0]

# --- results ---
print("From the equation:")
display(Math(rf"{sp.latex(eq_vel_sq)}"))
print("Solve for a:")
display(Math(rf"a = {sp.latex(a_expr)}"))
```

```python
# --- 1. Create Lambdified Functions ---
calc_a = sp.lambdify((v, v0, x), a_expr)

# --- 2. Astropy Physical Inputs ---
x_qty = 100 * u.m
v_qty = (0 * u.imperial.mi / u.h).to(u.m / u.s)
v0_qty = (60 * u.imperial.mi / u.h).to(u.m / u.s)

# --- 3. Numerical Calculations ---
a_qty = calc_a(v_qty, v0_qty, x_qty)[0]


# --- 4. Results ---
print(f"Required acceleration: {a_qty:,.2g}")
print(f"The minus sign means that is decceleration (breaking).")
```

#### Problem 2

A rocket is launched straight up and has an initial speed of 50 m/s. It has a constant upward acceleration of $ 2.00 m/s^2$ until the engines stop at an altitude of 150 m.

a) What is the rocket speed at engine cutoff?

b) How high does the rocket go?

<img src="../pics/Phy1_video_2_2.png" width=400> 

```python
# --- SymPy symbol vars ---
a, v, v0, y = sp.symbols("a, v, v_0, y", real=True, positive=True)

# --- Part a ---
print("--- Part a ---")
# --- kinematic equation ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * y)
v_expr = sp.solve(eq_vel_sq, v)[0]

# --- results ---
print("From the equation:")
display(Math(rf"{sp.latex(eq_vel_sq)}"))
print("Solve for v:")
display(Math(rf"v = {sp.latex(v_expr)}"))

# --- Part b ---
print("--- Part b ---")
y_expr = sp.solve(eq_vel_sq, y)[0]

# --- results ---
print("From the equation:")
display(Math(rf"{sp.latex(eq_vel_sq)}"))
print("Solve for y:")
display(Math(rf"y = {sp.latex(y_expr)}"))
```

```python
# --- 1. Create Lambdified Functions ---
calc_v = sp.lambdify((a, y, v0), v_expr)
calc_y = sp.lambdify((v, v0, a), y_expr)


# --- 2. Astropy Physical Inputs ---
y_cutoff_qty = 150 * u.m
v0_qty = 50 * (u.m / u.s)
a_qty = 2 * (u.m / u.s**2)
v_apex_qty = 0 * (u.m / u.s)
g_qty = -const.g0

# --- 3. Numerical Calculations ---
v_cuttoff_qty = calc_v(a_qty, y_cutoff_qty, v0_qty)

# At apex (max height) v=0, y2 from cuttoff to apex
y2_qty = calc_y(v_apex_qty, v_cuttoff_qty, g_qty)

# Total height
h_qty = y_cutoff_qty + y2_qty

# --- 4. Results ---
print(f" Part (a) - Velocity at MECO: {v_cuttoff_qty:,.1f}")
print(f" Part (b) - Total rocket height: {h_qty:,.1f}")
```
