#### Problem 1


Two people pull on a mule, as seen from above.

a) What is the single force that is equivalent to the two forces given?

b) What force would a third person have to exert to make the net force on the mule equal to zero?

<img src="../pics/Phy1_video_3_2.png" width=600> 

```python
# Solution using NumPy
# 1. Define force magnitudes and angles in degrees
F1_mag = 120  # N
theta1_deg = 60

F2_mag = 80  # N
theta2_deg = 180 - 75  # 105 degrees from +x-axis

# Convert angles from degrees to radians
theta1_rad = np.radians(theta1_deg)
theta2_rad = np.radians(theta2_deg)

# 2. Construct 2D vectors [Fx, Fy]
F1 = np.array([F1_mag * np.cos(theta1_rad), F1_mag * np.sin(theta1_rad)])
F2 = np.array([F2_mag * np.cos(theta2_rad), F2_mag * np.sin(theta2_rad)])

# --- Part (a): Equivalent (Net) Force ---
F_net = F1 + F2

# Magnitude using Euclidean norm
F_net_mag = np.linalg.norm(F_net)

# Ddirection angle using arctan2 (returns angle in radians)
angle_net_deg = np.degrees(np.arctan2(F_net[1], F_net[0]))

print("--- Part (a) ---")
print(f"Net Force Vector (Fx, Fy): [{F_net[0]:.0f}, {F_net[1]:.0f}] N")
print(f"Magnitude                : {F_net_mag:.0f} N")
print(f"Direction                : {angle_net_deg:.0f}° counterclockwise from +x axis")

# --- Part (b): Third Force for Equilibrium ---
F3 = -F_net

F3_mag = np.linalg.norm(F3)
angle_F3_deg = np.degrees(np.arctan2(F3[1], F3[0])) % 360

print()
print("--- Part (b) ---")
print(f"Force 3 Vector (Fx, Fy)  : [{F3[0]:.0f}, {F3[1]:.0f}] N")
print(f"Magnitude                : {F3_mag:.0f} N")
print(f"Direction                : {angle_F3_deg:.0f}° counterclockwise from +x axis")
```
