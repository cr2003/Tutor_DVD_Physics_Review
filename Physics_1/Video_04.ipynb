##  Projectile Motion

### x-dir

$$v_x = v_{x0} = v_0 \cos \theta$$

$$x = v_{x0} t = (v_0 \cos \theta) t$$

### y-dir

$$v_y = v_{y0} -gt $$

$$y = v_{y0} t - \frac{1}{2} g t^2$$

$$v_{y}^2 = v_{y0}^2 -2gy $$



$$A_x = A \cos \theta$$

$$A_y = A \sin \theta$$

$$A_x^2 + A_y^2 = A^2 \implies A = \sqrt{A_x^2 + A_y^2}$$

$$\tan \theta = \frac{A_y}{A_x}$$

#### Problem 1

A football is kicked 36 m from a field goal, which is 3.05 m high. The ball is kicked at a speed of 20 m/s at an angle of 53 degrees from the ground.

**a)** How much does the ball clear or fall short of the goal?

**b)** Is the ball rising or falling when it approaches the field goal?

<img src="../pics/Phy1_video_4.png" width=600> 

```python
# =====================================================================
# 1. SYMBOLIC STAGE (SymPy) - Calculate t to reach the goal
# =====================================================================

# Define abstract mathematical symbols
(
    x,
    v0,
    theta,
    t,
    a,
) = sp.symbols("x, v_0, theta, t, a", positive=True)

# Define the kinematic equation x = v0 * cos(theta) * t
eq_x = sp.Eq(x, v0 * sp.cos(theta) * t)

# Solve the equation for the unknown time 't'
solutions = sp.solve(eq_x, t)
t_expr = solutions[0]


print("--- time Formula Solved by SymPy ---")
display(Math(rf"{sp.latex(sp.Eq(t, t_expr))}"))
```

```python
# =====================================================================
# 2. UNITS AND NUMERICAL STAGE (Pure AstroPy + NumPy)
# =====================================================================

# --- 1. Derive callables from SymPy ---
calc_t = sp.lambdify((x, v0, theta), t_expr)

#  --- 2. Astropy Quantities (_qty suffix) ---
v0_qty = 20 * (u.m / u.s)
x_qty = 36 * u.m
theta_deg_qty = 53 * u.deg
theta_rad_qty = theta_deg_qty.to(u.rad)

# --- 3. calculation ---
t_qty = calc_t(x_qty, v0_qty, theta_rad_qty.value)


# --- 4. print results ---
print("--- a) t result ---")
print(f"Time to reach the goal: {t_qty:.0f}")
```

```python
# =====================================================================
# 3. UNITS AND NUMERICAL STAGE (Pure AstroPy + SymPy)
# With the previous time, calculate the high y
# =====================================================================

# 1. Define your inputs with AstroPy units
a = -const.g0
goal_y_qty = 3.05 * (u.m)

# 2. Calculate directly.
# kinematic equation: y = v0 * sin(theta) * t + 1/2 a t^2
y_qty = v0_qty * sp.sin(theta_rad_qty.value) * t_qty + 0.5 * a * t_qty**2


print("--- a) y result ---")
print(f"Ball high at crossing goal: {y_qty:.1f}")
print(f"The ball clears the goal by : {y_qty - goal_y_qty:.2f}")
```

```python
# =====================================================================
# 4. UNITS AND NUMERICAL STAGE (Pure AstroPy + SymPy)
# Calculate the vertical velocity of the ball when reaches the goal (v_y)
# =====================================================================

# kinematic equation: vy = v0 sin(theta) + at
vy_qty = v0_qty * sp.sin(theta_rad_qty.value) + a * t_qty

print("--- b) vy result ---")
print(f"Vertical velocity of the Ball at crossing goal: {vy_qty:.1f}")
print(f"The - sign indicates the ball is going down")
```

#### Problem

You throw a ball straight upward and it returns to your hand after 3 s in the air. A second ball is thrown at an angle of 30 degrees. What speed must the second ball be thrown so that it reaches the same height as the first ball?

<img src="../pics/Phy1_video_4_2.png" width=600> 

```python
# =====================================================================
# 1. UNITS AND NUMERICAL STAGE (Pure AstroPy + SymPy)
# The first ball its up in the air 3s. 1.5 up and 1.5 down.
# In the down, it starts at vy0 = 0, so we calculate how much its
# =====================================================================

# 1. Astropy Quantities (_qty suffix) ---
t_down_qty = 1.5 * u.s
a_qty = -const.g0
# kinematic equation: y = 0 + 1/2 a t^2
y_qty = 0.5 * a_qty * t_down_qty**2

print("--- a) y result ---")
print(f"Displacement of the ball: {y_qty:.2f}")
print(f"The - sign indicates the down fall, the heigth is {-y_qty:.2f}")
```

```python
# =====================================================================
# 2. SYMBOLIC STAGE (SymPy) - Calculate v0
# =====================================================================

# Define abstract mathematical symbols
x, y, v0, theta, a, zero = sp.symbols("x, y, v_0, theta, a, zero", positive=True)

# Define the kinematic equation: 0 = (v_0 * sin(theta))^2 + 2ay
eq_vel_sq = sp.Eq(zero, (v0 * sp.sin(theta)) ** 2 + 2 * a * y)

# Solve the equation for the unknown v_0
solutions = sp.solve(eq_vel_sq, v0)
v0_expr = solutions[1]

print("--- v0 Formula Solved by SymPy ---")
display(Math(rf"{sp.latex(sp.Eq(v0, v0_expr))}"))
```

```python
# =====================================================================
# 3. UNITS AND NUMERICAL STAGE (Pure AstroPy + SymPy)
# The second ball its thrown with an initial velocity v0 and in 1.5 s. reaches
# its max high.
# =====================================================================

# --- 1. Astropy Quantities (_qty suffix) ---
a_qty = -const.g0
y_high_qty = -y_qty
theta_deg_qty = 30 * u.deg
theta_rad_qty = theta_deg_qty.to(u.rad)

# --- 2. Derive callables from SymPy ---
calc_v0 = sp.lambdify((a, y, zero, theta), v0_expr)

v0_qty = calc_v0(a_qty, y_high_qty, 0, theta_rad_qty.value)

print("--- b) v0 result ---")
print(f"The second ball initial velocity: {v0_qty:.1f}")
```
