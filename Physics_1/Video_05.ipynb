##  Newtons Laws of Motion


### 1. The First Law: The Law of Inertia

An object at rest stays at rest, and an object in motion stays in motion with the same speed and in the same direction, unless acted upon by an unbalanced external force.

### 2. The Second Law: The Law of Force and Acceleration

The acceleration of an object depends directly upon the net force acting upon the object and inversely upon the mass of the object.

$$F = m a $$

Newton Units:
$$(kg)(\frac{m}{s^2}) = N$$

### 3. The Third Law: Action and Reaction

For every action, there is an equal and opposite reaction.

Forces always occur in pairs. If Object A exerts a force on Object B ($\vec{F}_{AB}$), Object B simultaneously exerts an equal and opposite force back on Object A ($\vec{F}_{BA}$).

#### Problem 1

A large mass of snow and ice breaks free from high on a mountaintop and starts a long frictionless ride to the ground. The mountain has a slope of 30 degrees and the mass of ice is initially 400 m up from the mountain base. If you are standing at the bottom, how long do you have to get out of the way?

<img src="../pics/Phy1_video_5.png" width=800> 

```python
# =====================================================================
# 1. SYMBOLIC STAGE (SymPy)
# =====================================================================

# Define abstract mathematical symbols
(x, v0, t, a, m, g, theta) = sp.symbols("x, v0, t, a, m, g, theta", positive=True)

# --- From the picture ---
# w sin(theta) = m g sin(theta) and F = m a
# so the equation: m g sin(theta) = ma
eq_a = sp.Eq(m * g * sp.sin(theta), m * a)

# Solve the equation for the unknown acceleration 'a'
solutions = sp.solve(eq_a, a)
a_expr = solutions[0]

print("--- acceleration Formula Solved by SymPy ---")
display(Math(rf"{sp.latex(sp.Eq(a, a_expr))}"))

# Define the kinematic equation: x = v0 t + 1/2 a t^2 (v0=0)
eq_x = sp.Eq(x, sp.Rational(1, 2) * a * t**2)

# Solve the equation for the unknown time 't'
solutions = sp.solve(eq_x, t)
t_expr = solutions[0]

print("--- time Formula Solved by SymPy ---")
display(Math(rf"{sp.latex(sp.Eq(t, t_expr))}"))
```

```python
# =====================================================================
# 2. UNITS AND NUMERICAL STAGE (Pure AstroPy + SymPy)
# =====================================================================

# --- 1. Derive callables from SymPy ---
calc_a = sp.lambdify((g, theta), a_expr)
calc_t = sp.lambdify((x, a), t_expr)


# --- 2. Astropy Quantities (_qty suffix) ---
x_qty = 400 * u.m
theta_deg_qty = 30 * u.deg
theta_rad_qty = theta_deg_qty.to(u.rad)
g_qty = const.g0

a_qty = calc_a(g_qty, theta_rad_qty.value)

t_qty = calc_t(x_qty, a_qty)

print("--- t result ---")
print(f"Time to get out: {t_qty:.1f}")
```

#### Problem 2

A 5 g bullet leaves the muzzle of a riffle with a speed of 320 m/s. The length of the barrel is 0.82 m. What force is exerted on the bullet while it is traveling down the barrel?  

<img src="../pics/Phy1_video_5_2.png" width=800> 

```python
# =====================================================================
# 1. SYMBOLIC STAGE (SymPy)
# =====================================================================

# Define abstract mathematical symbols
(
    x,
    v,
    v0,
    a,
) = sp.symbols("x, v, v0, a", positive=True)

# Define the kinematic equation: v^2 = v_0^2 + 2 a x
# Since the initial velocity is zero, we have
equation = sp.Eq(v**2, 2 * a * x)

# Solve the equation for the unknown 'a'
solutions = sp.solve(equation, a)
a_expr = solutions[0]

print("--- time Formula Solved by SymPy ---")
display(Math(rf"{sp.latex(sp.Eq(a, a_expr))}"))
```

```python
# =====================================================================
# 2. UNITS AND NUMERICAL STAGE (Pure AstroPy + SymPy)
# =====================================================================

# --- 1. Derive callables from SymPy ---
calc_a = sp.lambdify((v, x), a_expr)

# 2. Define your inputs with AstroPy units
x_qty = 0.82 * u.m
m_qty = (5 * u.g).to(u.kg)
v_qty = 320 * (u.m / u.s)

a_qty = calc_a(v_qty, x_qty)

# equation F = m a
F_qty = m_qty * a_qty

print("--- F result ---")
print(f"Force: {F_qty.to(u.N):.0f}")
```

#### Problem 3

Find the tension in the two wires that support the light fixture. The light wighs 100 N.  

<img src="../pics/Phy1_video_5_3.png" width=800> 


```python
# =====================================================================
# 1. SYMBOLIC STAGE (SymPy)
# =====================================================================

# Define abstract mathematical symbols
(
    T_1,
    theta,
    w,
) = sp.symbols("T_1, theta, w", positive=True)

# Since the tension in the wires is the same, T_1 = T_2
# Define the Equilibrium equation: T_1 sin(theta) + T_1 sin(theta) - 100 = 0

eq_equilib = sp.Eq(0, T_1 * sp.sin(theta) + T_1 * sp.sin(theta) - w)

# Solve the equation for the unknown T_1
solutions = sp.solve(eq_equilib, T_1)
T_1_expr = solutions[0]

print("--- Tension Formula Solved by SymPy ---")
display(Math(rf"{sp.latex(sp.Eq(T_1, T_1_expr))}"))
```

```python
# =====================================================================
# 2. UNITS AND NUMERICAL STAGE (Pure AstroPy + Sympy)
# =====================================================================

# --- 1. Derive callables from SymPy ---
calc_T = sp.lambdify((w, theta), T_1_expr)


# 2. Define your inputs with AstroPy units
w_qty = 100 * u.N
theta_deg_qty = 40 * u.deg
theta_rad_qty = theta_deg_qty.to(u.rad)

T1_qty = calc_T(w_qty, theta_rad_qty.value)

print("--- Wire Tension result ---")
print(f"Wire tension: {T1_qty:.0f}")
```

#### Problem 4

A 2,000 kg sailboat has an 3,000 N eastward force exerted on it by the ocean waves. The wind also blows against the sails with a force of 6,000 N, directed 45 degrees north of west. What is the magnitude of the resultant acceleration?

<img src="../pics/Phy1_video_5_4.png" width=800> 

```python
# =====================================================================
# UNITS AND NUMERICAL STAGE (Pure AstroPy + NumPy)
# =====================================================================

# 1. Define your inputs with AstroPy units
m_qty = 2000 * u.kg
F1_qty = 3000 * u.N
F2_qty = 6000 * u.N
theta_qty = 45 * u.deg

# NOTE: NumPy takes care of the magnitude in deg and do the neccesary conversions
Fx_qty = -F2_qty * np.cos(theta_qty) + F1_qty
Fy_qty = F2_qty * np.sin(theta_qty)
F_mag_qty = np.sqrt(Fx_qty**2 + Fy_qty**2)

# F = m a ==> a = F / m
a_qty = F_mag_qty / m_qty

print(f"Magnitude resultant acceleration: {a_qty.to(u.m / u.s**2):.1f}")
```
